```python
import requests
import pandas as pd
```

```python
BASE_URL = "https://api.data.amsterdam.nl/v1/peilmerken/"
BRUG_KEYWORDS = ["brug", "viaduct", "ophaalbrug", "hefbrug", "spoorbrug"]

def fetch_peilmerken():
    peilmerken = []
    url = BASE_URL
    while url:
        print(f"Fetching: {url}")
        response = requests.get(url)
        data = response.json()
        
        embedded = data.get("_embedded", {})
        items = embedded.get("peilmerken", [])

        for item in items:
            omschrijving = item.get("omschrijving", "").lower()
            if any(keyword in omschrijving for keyword in BRUG_KEYWORDS):
                peilmerken.append({
                    "id": item.get("code"),
                    "omschrijving": omschrijving,
                    "hoogte_tov_nap_m": item.get("hoogte_nap"),
                    "coordinaten": item.get("punt", {}).get("coordinates"),
                    "datum_opmeting": item.get("datum_opmeting")
                })
        
        # Pagination
        url = data.get("_links", {}).get("next", {}).get("href")

    return pd.DataFrame(peilmerken)

# Run the script
df = fetch_peilmerken()
df.to_csv("amsterdam_brug_peilmerken.csv", index=False)
print(df.head())
```
